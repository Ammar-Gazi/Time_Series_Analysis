```python
# %% [1] Configuration & Imports
import os
import json
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.tsa.statespace.sarimax import SARIMAX
from sklearn.metrics import mean_absolute_error, mean_squared_error

warnings.filterwarnings("ignore")
plt.style.use("seaborn-v0_8-whitegrid")

CONFIG = {
    # RapidAPI Settings (Uses local cached JSON fallback if RAPIDAPI_KEY is empty)
    "imdb_id": "tt4154796",  # Avengers: Endgame
    "rapidapi_key": os.getenv("RAPIDAPI_KEY", ""),
    "rapidapi_host": "imdb236.p.rapidapi.com",
    "cache_dir": Path("data"),
    
    # Time Series & Modeling Settings
    "run_days": 63,              # 9 full weeks (63 days) of theatrical run
    "seasonal_period": 7,        # 7-day weekly box office cycle (Fri-Sun surge)
    "train_ratio": 0.78,         # First ~49 days train, last 14 days (2 weeks) test
    "arima_order": (1, 1, 1),    # (p, d, q)
    "seasonal_order": (1, 0, 1, 7),  # (P, D, Q, s=7) weekly seasonality
    
    # Studio Business Economics
    "studio_split_pct": 0.52,    # Studio keeps ~52% of gross box office
    "marketing_multiplier": 0.50,# Prints & Advertising (P&A) ~50% of production budget
    "daily_overhead_per_theater": 650.0  # Minimum daily gross/theater to keep screens
}

CONFIG["cache_dir"].mkdir(parents=True, exist_ok=True)
print("✅ Block [1] Configuration loaded successfully.")


# %% [2] RapidAPI Movie Ingestion & JSON Flattener
SAMPLE_RAPIDAPI_JSON = {
    "id": "tt4154796",
    "title": "Avengers: Endgame",
    "year": 2019,
    "release_date": "2019-04-26",
    "runtime_minutes": 181,
    "certificate": "PG-13",
    "genres": ["Action", "Adventure", "Sci-Fi"],
    "rating": {"value": 8.4, "vote_count": 1500450, "top_rank": 74},
    "metascore": {"score": 78, "review_count": 57},
    "directors": [{"name": "Anthony Russo"}, {"name": "Joe Russo"}],
    "stars": [
        {"name": "Robert Downey Jr."},
        {"name": "Chris Evans"},
        {"name": "Mark Ruffalo"},
        {"name": "Chris Hemsworth"}
    ],
    "total_cast": 210,
    "image": {"link": "https://m.media-amazon.com/images/M/MV5BMTc5MDE2ODcwNV5BMl5BanBnXkFtZTgwMzI2NzQ2NzM@._V1_.jpg"},
    "box_office": {
        "budget": {"amount": 356000000, "currency": "USD"},
        "opening_weekend": {
            "amount": 357115007,
            "currency": "USD",
            "weekend_end_date": "2019-04-28",
            "theater_count": 4662
        },
        "gross_worldwide": {"amount": 2938371784, "currency": "USD"}
    },
    "similar_titles": [
        {"id": "tt4154756", "title": "Avengers: Infinity War", "year": 2018, "rating": {"value": 8.4}, "is_series": False},
        {"id": "tt0848228", "title": "The Avengers", "year": 2012, "rating": {"value": 8.0}, "is_series": False},
        {"id": "tt2395427", "title": "Avengers: Age of Ultron", "year": 2015, "rating": {"value": 7.3}, "is_series": False},
        {"id": "tt10872600", "title": "Spider-Man: No Way Home", "year": 2021, "rating": {"value": 8.1}, "is_series": False},
        {"id": "tt0468569", "title": "The Dark Knight", "year": 2008, "rating": {"value": 9.1}, "is_series": False}
    ]
}

def parse_rapidapi_movie(payload: dict) -> dict:
    """Flattens nested RapidAPI IMDb JSON and calculates studio unit economics."""
    box = payload.get("box_office") or {}
    budget_amt = (box.get("budget") or {}).get("amount") or 0
    ow_block = box.get("opening_weekend") or {}
    ow_amt = ow_block.get("amount") or 0
    theaters = ow_block.get("theater_count") or 1
    ww_gross = (box.get("gross_worldwide") or {}).get("amount") or 0

    rating_block = payload.get("rating") or {}
    meta_block = payload.get("metascore") or {}

    # Studio Break-Even Math: Total Cost = Budget * (1 + Marketing%) / Studio Split%
    total_cost_usd = budget_amt * (1 + CONFIG["marketing_multiplier"])
    break_even_box_office = total_cost_usd / CONFIG["studio_split_pct"]

    return {
        "imdb_id": payload.get("id"),
        "title": payload.get("title"),
        "release_date": payload.get("release_date"),
        "runtime_minutes": payload.get("runtime_minutes"),
        "certificate": payload.get("certificate"),
        "genres": ", ".join(payload.get("genres") or []),
        "imdb_rating": rating_block.get("value", 7.0),
        "vote_count": rating_block.get("vote_count", 0),
        "metascore": meta_block.get("score", 65),
        "budget_usd": budget_amt,
        "opening_weekend_usd": ow_amt,
        "theater_count": theaters,
        "gross_worldwide_usd": ww_gross,
        "per_theater_avg_ow": round(ow_amt / theaters, 2),
        "legs_multiplier": round(ww_gross / ow_amt, 2) if ow_amt else np.nan,
        "break_even_gross_usd": round(break_even_box_office, 2),
        "poster_url": (payload.get("image") or {}).get("link"),
        "similar_ids": [
            m["id"] for m in (payload.get("similar_titles") or [])
            if not m.get("is_series", False)
        ]
    }

movie_meta = parse_rapidapi_movie(SAMPLE_RAPIDAPI_JSON)
print(f"🎬 Loaded Movie: {movie_meta['title']} ({movie_meta['release_date']})")
print(f"   • Production Budget:      ${movie_meta['budget_usd']:,.0f}")
print(f"   • Opening Weekend (Dom):  ${movie_meta['opening_weekend_usd']:,.0f} across {movie_meta['theater_count']:,} theaters")
print(f"   • Studio Break-Even Goal: ${movie_meta['break_even_gross_usd']:,.0f}")


# %% [3] Reconstruct & Preprocess Daily Theatrical Time Series (Day 1 to Day 63)
def build_theatrical_timeseries(meta: dict, days: int = 63, seed: int = 42) -> pd.DataFrame:
    """
    Constructs a realistic contiguous daily box office time series anchored strictly
    to the RapidAPI movie's release_date, opening_weekend_usd, theater_count, and ratings.
    """
    rng = np.random.default_rng(seed)
    start_date = pd.to_datetime(meta["release_date"])
    date_index = pd.date_range(start=start_date, periods=days, freq="D")

    # Word-of-mouth decay rate calibrated by IMDb rating & Metascore
    quality_score = (meta["imdb_rating"] / 10.0) * 0.6 + (meta["metascore"] / 100.0) * 0.4
    daily_decay_rate = 0.082 - (0.025 * quality_score)  # Higher quality = slower decay

    # Day-of-week theatrical multipliers (Fri=4, Sat=5, Sun=6, Tue=1 discount bump)
    dow_weights = {
        0: 0.28,  # Monday drop
        1: 0.38,  # Discount Tuesday bump
        2: 0.24,  # Wednesday
        3: 0.23,  # Thursday
        4: 0.95,  # Friday surge
        5: 1.18,  # Saturday peak
        6: 0.87   # Sunday matinee
    }

    # Calibrate initial scale so Days 1-3 (Fri, Sat, Sun) sum to exact API opening_weekend_usd
    ow_weights_sum = dow_weights[4] + dow_weights[5] + dow_weights[6]
    base_friday_scale = meta["opening_weekend_usd"] / ow_weights_sum

    daily_gross = []
    active_theaters = []
    current_theaters = meta["theater_count"]

    for t, dt in enumerate(date_index):
        dow = dt.dayofweek
        # Theater count steps down each Friday after Week 2
        if t >= 14 and dow == 4:
            current_theaters = int(current_theaters * 0.84)

        if t < 3:
            # Exact Opening Weekend calibration
            gross = base_friday_scale * dow_weights[dow] * (1 + rng.normal(0, 0.015))
        else:
            # Exponential decay + 7-day weekly seasonality + noise
            decay_factor = np.exp(-daily_decay_rate * (t - 1))
            seasonal_factor = dow_weights[dow]
            noise = rng.normal(1.0, 0.04)
            gross = base_friday_scale * decay_factor * seasonal_factor * noise

        daily_gross.append(max(gross, 50_000.0))
        active_theaters.append(max(current_theaters, 250))

    df = pd.DataFrame({
        "date": date_index,
        "day_of_run": np.arange(1, days + 1),
        "daily_gross_usd": daily_gross,
        "active_theaters": active_theaters,
    }).set_index("date").asfreq("D")

    # Rescale Days 1-3 to match exact RapidAPI opening_weekend_usd to the dollar
    ow_actual_sum = df.iloc[0:3]["daily_gross_usd"].sum()
    df.iloc[0:3, df.columns.get_loc("daily_gross_usd")] *= (meta["opening_weekend_usd"] / ow_actual_sum)

    # Feature Engineering for Exogenous Modeling
    df["is_weekend"] = df.index.dayofweek.isin([4, 5, 6]).astype(int)
    df["is_discount_tuesday"] = (df.index.dayofweek == 1).astype(int)
    df["log_daily_gross"] = np.log1p(df["daily_gross_usd"])
    df["cumulative_dom_gross"] = df["daily_gross_usd"].cumsum()
    df["per_theater_daily_usd"] = df["daily_gross_usd"] / df["active_theaters"]

    return df

ts_df = build_theatrical_timeseries(movie_meta, days=CONFIG["run_days"])
ts_df.to_csv(CONFIG["cache_dir"] / f"{movie_meta['imdb_id']}_daily_timeseries.csv")
print(f"✅ Block [3] Daily Time Series built: {len(ts_df)} contiguous days.")
print(f"   • Verified Opening Weekend Sum (Days 1-3): ${ts_df.iloc[:3]['daily_gross_usd'].sum():,.0f}")
display(ts_df.head(7)[["day_of_run", "daily_gross_usd", "active_theaters", "per_theater_daily_usd", "is_weekend"]])


# %% [4] Time Series Decomposition (Trend, 7-Day Weekly Seasonality, Residuals)
decomp = seasonal_decompose(
    ts_df["log_daily_gross"],
    model="additive",
    period=CONFIG["seasonal_period"]
)

fig, axes = plt.subplots(4, 1, figsize=(11, 8), sharex=True)
decomp.observed.plot(ax=axes[0], color="#0066FF", lw=2, title=f"{movie_meta['title']} — Log Daily Box Office (Observed)")
decomp.trend.plot(ax=axes[1], color="#00C2FF", lw=2, title="Underlying Theatrical Decay Trend")
decomp.seasonal.plot(ax=axes[2], color="#2563EB", lw=1.5, title="7-Day Weekly Moviegoing Cycle (Fri-Sun Peaks, Tue Bumps)")
decomp.resid.plot(ax=axes[3], color="#64748B", style="o", ms=4, title="Residuals (Shocks / Anomalies)")
plt.tight_layout()
plt.show()


# %% [5] Chronological Train-Test Split & SARIMAX Modeling
split_idx = int(len(ts_df) * CONFIG["train_ratio"])
train_df = ts_df.iloc[:split_idx].copy()
test_df = ts_df.iloc[split_idx:].copy()

exog_cols = ["is_weekend", "is_discount_tuesday"]

# Fit SARIMAX on log_daily_gross so variance stays stabilized across exponential decay
model = SARIMAX(
    endog=train_df["log_daily_gross"],
    exog=train_df[exog_cols],
    order=CONFIG["arima_order"],
    seasonal_order=CONFIG["seasonal_order"],
    enforce_stationarity=False,
    enforce_invertibility=False
)
fitted_model = model.fit(disp=False)

# Forecast over the test window with 95% confidence intervals
forecast_res = fitted_model.get_forecast(steps=len(test_df), exog=test_df[exog_cols])
test_df["pred_log_gross"] = forecast_res.predicted_mean
test_df["pred_daily_gross_usd"] = np.expm1(test_df["pred_log_gross"])

conf_int = forecast_res.conf_int(alpha=0.05)
test_df["lower_ci_usd"] = np.expm1(conf_int.iloc[:, 0])
test_df["upper_ci_usd"] = np.expm1(conf_int.iloc[:, 1])

print(f"✅ Block [5] SARIMAX{CONFIG['arima_order']}x{CONFIG['seasonal_order']} trained on {len(train_df)} days, forecasted {len(test_df)} days.")


# %% [6] Evaluation Metrics (MAE, RMSE, MAPE) & Forecast Visualization
def evaluate_forecast(y_true: pd.Series, y_pred: pd.Series) -> dict:
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mape = np.mean(np.abs((y_true - y_pred) / np.maximum(y_true, 1.0))) * 100
    return {"MAE ($)": mae, "RMSE ($)": rmse, "MAPE (%)": mape}

metrics = evaluate_forecast(test_df["daily_gross_usd"], test_df["pred_daily_gross_usd"])
print("📊 Out-of-Sample Test Set Performance (Days 50 to 63):")
for k, v in metrics.items():
    print(f"   • {k:10s}: {v:,.2f}")

# Plot Train, Actual Test, and Forecast with 95% Confidence Interval (Skip Week 1 mega-spike for zoom clarity)
plot_train = train_df.iloc[7:]
fig, ax = plt.subplots(figsize=(11, 5))
ax.plot(plot_train.index, plot_train["daily_gross_usd"] / 1e6, label="Historical Daily Gross (Week 2-7)", color="#1E293B", lw=1.8)
ax.plot(test_df.index, test_df["daily_gross_usd"] / 1e6, label="Actual Test Gross", color="#0066FF", lw=2.2)
ax.plot(test_df.index, test_df["pred_daily_gross_usd"] / 1e6, label=f"SARIMAX Forecast (MAPE: {metrics['MAPE (%)']:.1f}%)", color="#00C2FF", ls="--", lw=2.2)
ax.fill_between(
    test_df.index,
    test_df["lower_ci_usd"] / 1e6,
    test_df["upper_ci_usd"] / 1e6,
    color="#00C2FF",
    alpha=0.18,
    label="95% Confidence Band"
)
ax.set_title(f"{movie_meta['title']} — Late-Run Box Office Forecast & OTT Transition Window", fontsize=13, fontweight="bold")
ax.set_ylabel("Daily Box Office Gross ($ Millions)")
ax.legend(frameon=True)
plt.tight_layout()
plt.show()


# %% [7] Business Decision Layer: Screen Allocation & Optimal OTT Streaming Transition Date
test_df["pred_per_theater_usd"] = test_df["pred_daily_gross_usd"] / test_df["active_theaters"]
below_overhead = test_df[test_df["pred_per_theater_usd"] < CONFIG["daily_overhead_per_theater"]]

print("\n💼 STUDIO EXECUTIVE & THEATER OPERATIONS RECOMMENDATION:")
print(f"   • Worldwide Box Office vs. Break-Even: ${movie_meta['gross_worldwide_usd']:,.0f} vs. ${movie_meta['break_even_gross_usd']:,.0f} ({movie_meta['roi_multiple'] if 'roi_multiple' in movie_meta else movie_meta['gross_worldwide_usd']/movie_meta['budget_usd']:.2f}x Budget)")

if not below_overhead.empty:
    ott_switch_date = below_overhead.index[0].strftime("%Y-%m-%d")
    switch_day_num = int(below_overhead.iloc[0]["day_of_run"])
    print(f"   • Recommended PVOD / OTT Streaming Launch Date: {ott_switch_date} (Day {switch_day_num} of release)")
    print(f"     Reason: Forecasted weekday per-theater gross drops below ${CONFIG['daily_overhead_per_theater']:,.0f}/day threshold.")
else:
    print("   • Hold Theatrical Exclusivity: Per-theater daily gross remains profitable above overhead across entire 63-day window.")
```
